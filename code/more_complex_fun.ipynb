## Sine

### explore a number of frequencies $(1, 10, 100, 1\_000, 10\_000)$






````python
# sine() takes an extra num_crossings argument, but every place that uses
# fun/f (sample_y, sample_y_bernoulli, explore_data, main_loop,
# animate_learned_function) calls it as a plain 1-argument fun(x) — same
# contract centered_parabola already satisfies. so instead of threading
# num_crossings through all of those, bind it once with functools.partial
# (same trick already used for MLPClassifier's fixed hyperparameters below),
# and pass the resulting 1-argument callable everywhere fun/f is expected.
from functools import partial

# NUMBER OF CROSSINGS of the x axis 
'''
Remark: num_crossings (hence m) is naively the number of crossings I want to see in the open interval (0, 10), hence if I want all points to the left of 5 being positive and all those to the right of 5 being negative I should set m = 1!
'''
m = 2
myfun = partial(sine, num_crossings=m)

# explore data distribution for this new f
explore_data(fun=myfun, n_train=50, supp_size=10_000, sigmas=(10**(-4), 1, 2, 5))
````

````python
# MODEL
mymodel = DecisionTreeClassifier

# PARAMETERS
N = 10_000
n_test = 200_000
n = 50
noise_std = 2
T = 100

# build test set
x_test = sample_x(n_test, N)
y_test = sample_y(x = x_test, f = myfun, sigma = noise_std)


# build models and do plots
hat_Rs, Rs, gaps, std_hat_Rs, std_Rs, std_gaps, predictions, x, y, x_grid = main_loop(x_test, y_test, n_train=n, sigma=noise_std, num_iter=T, model=mymodel, fun=myfun)
plots(hat_Rs, Rs, gaps, std_hat_Rs=std_hat_Rs, std_Rs=std_Rs, std_gaps=std_gaps, n_train=n, sigma=noise_std, num_iter=T, model=mymodel)
````

````python
# show animated function

animation = animate_learned_function(model = mymodel, n_train = n, best_pred = predictions,
                         best_x = x, best_y = y, x_grid = x_grid, fun = myfun)
HTML(animation.to_jshtml())
````

### per quale rapporto tra il numero di sign changes e num di training points ho test error = $0.5$? 



spoiler: per $r \approx 6.5$, che vuol dire, fissando $n = 50$,  $3250$ crossings. bah! 

````python
# find r_c: fix the ratio r = m/n (number of sine crossings over number of
# training points) and look for the threshold r_c above which the test error
# saturates at 0.5 (as good as guessing). known: 2 <= r_c <= 20, so sweep a
# handful of values spanning that range and compare the train/test error
# plots side by side.
ratios = [20, 15, 12, 10, 5, 2]

# MODEL
mymodel = DecisionTreeClassifier

# PARAMETERS (fixed across ratios, same as the earlier sine experiments)
N = 10_000
n_test = 200_000
n = 50
noise_std = 2
T = 100

fig, axs = plt.subplots(3, 2, figsize=(12, 14))
fig.suptitle(r'train vs test error for different $r = m/n$ ratios', fontsize=13)

for ax, r in zip(axs.flat, ratios):
    m = round(r * n)
    myfun = partial(sine, num_crossings=m)

    # build test set
    x_test = sample_x(n_test, N)
    y_test = sample_y(x=x_test, f=myfun, sigma=noise_std)

    # build models and do plots (only the train vs test error plot)
    hat_Rs, Rs, gaps, std_hat_Rs, std_Rs, std_gaps, *_ = main_loop(x_test, y_test, n_train=n, sigma=noise_std, num_iter=T, model=mymodel, fun=myfun)
    plots(hat_Rs, Rs, gaps, std_hat_Rs=std_hat_Rs, std_Rs=std_Rs, std_gaps=std_gaps, n_train=n, sigma=noise_std, num_iter=T, model=mymodel, two=False, three=False, axes=[ax])
    ax.set_title(f'r = {r}  (m = {m})\ntrain vs test error')

plt.tight_layout(rect=[0, 0, 1, 0.97])
plt.show()
````
